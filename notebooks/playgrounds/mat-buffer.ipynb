```python
import slangpy as spy
import numpy as np
```

```python
device = spy.create_device()
```

# Vector Buffer

```python
vec_src = '''
float3 vecFunc(int id){
    return float3(id);
}
'''
module = spy.Module.load_from_source(
    device,
    "vec_module",
    vec_src
)
module
```

```python
res_buf = spy.NDBuffer(device, dtype="float3", shape=(2,))
module.vecFunc(spy.call_id(), _result=res_buf)
res_buf.to_numpy()
```

```python
arr = np.random.rand(2, 3).astype(np.float32)
arr
```

```python
res_buf.copy_from_numpy(arr)
res_buf.to_numpy()
```

```python
cur = res_buf.cursor()
```

# High Dimensional Array Buffer

```python
arr_src = '''
float[3][3][3] arrFunc(int id){
    float[3][3][3] res;
    for (int i = 0; i < 3; i++)
        for (int j = 0; j < 3; j++)
            for (int k = 0; k < 3; k++)
                res[i][j][k] = id;
    return res;
}
'''
module = spy.Module.load_from_source(
    device,
    "arr_module",
    arr_src
)
module
```

```python
res_buf = spy.NDBuffer(device, dtype="float[3][3][3]", shape=(2,))
module.arrFunc(spy.call_id(), _result=res_buf)
res_buf.to_numpy()
```

```python
arr = np.random.rand(2, 3, 3, 3).astype(np.float32)
arr
```

```python
cur = res_buf.cursor()
```

```python
cur.element_type.col_count
```

```python
res_buf.copy_from_numpy(arr)
res_buf.to_numpy()
```
