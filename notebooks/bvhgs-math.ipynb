# Quaternions

## Multiplication

```python
# quaternion.mul
q1 = np.random.rand(8, 4).astype(np.float32)
q1 = q1 / np.linalg.norm(q1, axis=1, keepdims=True)
q2 = np.random.rand(8, 4).astype(np.float32)
q2 = q2 / np.linalg.norm(q2, axis=1, keepdims=True)
q_ref = quaternion.as_float_array(
    quaternion.as_quat_array(q1) * quaternion.as_quat_array(q2)
)
q_ref
```

```python
# San check: unit quaternion.
np.linalg.norm(q_ref, axis=1)
```

```python
# Convert from wxyz -> xyzw.
q1_spy = np.ascontiguousarray(q1[:, [1, 2, 3, 0]])
q2_spy = np.ascontiguousarray(q2[:, [1, 2, 3, 0]])
```

```python
# Dispatch quat.mul
quat_mul = math_module.find_function("quat.mul")
q3 = quat_mul(q1_spy, q2_spy, _result="numpy")
np.linalg.norm(q3 - q_ref[:, [1, 2, 3, 0]])
```
