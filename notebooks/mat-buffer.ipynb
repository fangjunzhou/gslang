# Matrix Buffer Issue

```python
import slangpy as spy
import numpy as np
```

```python
device = spy.create_device()
```

```python
mat_src = '''
float3x3 matFunc(int id){
    return float3x3(id);
}
'''
module = spy.Module.load_from_source(
    device,
    "mat_module",
    mat_src
)
module
```

```python
res_buf = spy.NDBuffer(device, dtype="float3x3", shape=(2,))
module.matFunc(spy.call_id(), _result=res_buf)
res_buf
```

```python
arr = res_buf.to_numpy()
arr
```

```python
arr = np.random.rand(2, 3, 3).astype(np.float32)
arr
```

```python
res_buf.copy_from_numpy(arr)
res_buf.to_numpy()
```

```python
arr_slice = np.random.rand(2, 4, 4)[:, :3, :3]
arr_slice
```

```python
res_buf.copy_from_numpy(arr_slice)
```
